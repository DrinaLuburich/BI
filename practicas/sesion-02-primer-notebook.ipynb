```python
#Tarea 2 · Python como calculadora
#Calcula cuántas líneas de ticket genera Colmena Market al día: 8 líneas por ticket, 1.000 tickets por tienda y 120 tiendas.
#Resultado esperado: 960000

linea_por_ticket = 8
ticket_por_tienda = 1000
num_tiendas= 120
print("Tarea 2")
print("Líneas de ticket por día: ",linea_por_ticket*ticket_por_tienda*num_tiendas)
```

```python
#Tarea 3 · Las cámaras de frío, con variables
#Cada tienda tiene 3 cámaras de frío, y cada cámara envía su temperatura 12 veces por hora, las 24 horas del día.
#Crea una variable para cada dato: tiendas, camaras_por_tienda y lecturas_por_hora.
#Calcula lecturas_al_dia a partir de ellas.
#Calcula lecturas_al_anio. Los sensores no descansan: son 365 días.
#Muestra los dos resultados con print y una f-string.
#Resultado esperado:
#Lecturas al día: 103680 // Lecturas al año: 37843200

print("Tarea 3")

num_tiendas = 120
num_camaras_por_tienda = 3
num_lecturas_por_hora = 12
num_lecturas_dia = 24
num_lecturas_anio = 365

print("Lecturas al día: ",num_tiendas*num_camaras_por_tienda*num_lecturas_por_hora*num_lecturas_dia)
print("Lecturas al año: ",num_tiendas*num_camaras_por_tienda*num_lecturas_por_hora*num_lecturas_dia*num_lecturas_anio)

#FORMAT
calc_lecturas_dia = num_tiendas*num_camaras_por_tienda*num_lecturas_por_hora*num_lecturas_dia
calc_lecturas_anio =  num_tiendas*num_camaras_por_tienda*num_lecturas_por_hora*num_lecturas_dia*num_lecturas_anio

print(f"Lecturas al día: {calc_lecturas_dia:.1f}")
print(f"Lecturas al día: {calc_lecturas_anio:.1f}")
```

```python
#Tarea 4 · De lecturas a megas
#Cada lectura ocupa unos 46 bytes. ¿Cuánto ocupan las lecturas de un día? ¿Y las de un año?
#Crea la variable bytes_por_lectura.
#Calcula megas_al_dia. Recuerda que 1 MB son 1000000 de bytes.
#Calcula gigas_al_anio. Recuerda que 1 GB son 1000 MB.
#Muestra los resultados con dos decimales, usando :.2f dentro de las llaves.
#Resultado esperado:
#Sensores: 4.77 MB al día // Sensores: 1.74 GB al año
print("Tarea 4")

num_byte_lecturas= 46
num_lecturas_dia = 24
num_lecturas_anio = 365

calc_lecturas_dia = num_tiendas*num_camaras_por_tienda*num_lecturas_por_hora*num_lecturas_dia
calc_lecturas_anio =  num_tiendas*num_camaras_por_tienda*num_lecturas_por_hora*num_lecturas_dia*num_lecturas_anio

calc_megas_al_dia= (num_byte_lecturas*calc_lecturas_dia) / 1000000
calc_gigas_al_anio= (num_byte_lecturas*calc_lecturas_anio) / (1000*1000000)


print(f"Megas al día: {calc_megas_al_dia:.2f}")
print(f"Megas al año: {calc_gigas_al_anio:.2f}")
```

```python
#Tarea 5 · Los clics de la web y la app
#La web y la app de Colmena Market reciben unas 250.000 visitas al día. En cada visita se registran unos 12 clics, y cada clic ocupa unos 220 bytes.
#Calcula, con variables, cuántos clics hay al día, cuántos megas ocupan y cuántos gigas son al año.
#Resultado esperado:
#Clics al día: 3,000,000
#Clics: 660 MB al día
#Clics: 240.9 GB al año

visitas_al_dia = 250000
clics_por_visita = 12
bytes_por_clic = 220

calc_clics_al_dia = visitas_al_dia*clics_por_visita
print(f"Clics al día: {calc_clics_al_dia:,}")

calc_megas_al_dia= (bytes_por_clic*calc_clics_al_dia) / 1000000
print(f"Megas al día: {calc_megas_al_dia:.0f} MB")

calc_gigas_al_anio= (calc_megas_al_dia*365)/1000
print("Gigas al año: ",calc_gigas_al_anio,"GB")


```
